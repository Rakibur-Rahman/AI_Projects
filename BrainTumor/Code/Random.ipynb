```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from PIL import Image
from sklearn.metrics import confusion_matrix
import seaborn as sns
```

```python
import os
path=os.listdir('/content/drive/MyDrive/BrainTumorClassificationProject/BrainTumorDataSet/Train')
classes={'Without_Tumor':0, 'Tumor':1}
```

```python
import cv2
X=[]
Y=[]
for cls in classes:
  pth='/content/drive/MyDrive/BrainTumorClassificationProject/BrainTumorDataSet/Train/'+cls
  for j in os.listdir(pth):
    img=cv2.imread(pth+'/'+j,0)
    img=cv2.resize(img,(200,200))
    X.append(img)
    Y.append(classes[cls])

```

```python
X=np.array(X)
Y=np.array(Y)
```

```python
np.unique(Y)
```

```python
pd.Series(Y).value_counts()
```

```python
X.shape
```

```python
plt.imshow(X[1],cmap='gray')
```

```python
X_updated=X.reshape(len(X),-1)
X_updated.shape
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(X_updated,Y,random_state=10,test_size=.20)
```

```python
print(xtrain.max(),xtrain.min())
print(xtest.max(),xtest.min())
xtrain=xtrain/255
xtest=xtest/255
print(xtrain.max(),xtrain.min())
print(xtest.max(),xtest.min())
```

```python
from sklearn.decomposition import PCA
```

```python
print(xtrain.shape,xtest.shape)
pca=PCA(.98)
pca_train=xtrain
pca_test=xtest
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
clf = RandomForestClassifier(n_estimators = 100)
```

```python
import warnings
warnings.filterwarnings('ignore')

lg=clf
lg.fit(pca_train, ytrain)
```

```python
print("Training Score:", lg.score(pca_train, ytrain))
print("Testning Score:", lg.score(pca_test, ytest))
```

```python
actual=ytest
predicted=y_pred
```
